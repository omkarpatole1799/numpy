```python
import pandas as pd
```

```python
df_olympics = pd.read_csv('data/bios_raw.csv')
df_olympics.sample(3)
```

```python
# make a copy of df
df = df_olympics.copy()
```

```python
# cleanup full name remove the '•' string
df["name"] = df['Full name'].str.replace('•',' ')
# df[["name", "Full name"]].sample(5)
```

```python
df.sample(4)

# df.info()
```

```python
# split measurements into height_cm and weight_kg
df[['height_cm', 'weight_kg']] = df["Measurements"].str.split('/', expand=True)
```

```python
# find total rows in which there is kg value in height_cm
# len(df[df["height_cm"].astype('str').str.contains("kg")])
# 1182

# 180 cm / 80 kg
# 76 kg

df['height_cm'] = pd.to_numeric(df['height_cm'].str.strip(" cm"), errors="coerce")

df['weight_kg'] = pd.to_numeric(df['weight_kg'].str.strip(' kg'), errors="coerce")

```

```python
# born and dates clearing
df.sample(5)
date_pattern = r"(\d{1,2}\s\w+\s\d{4}|\d{4})"
df["born_date"] = df["Born"].str.extract(date_pattern)
df["born_date"] = pd.to_datetime(df["born_date"], format="mixed", errors="coerce")
```

```python
df.head()
```

```python
# Extracting NOC from Born column
# df["NOC"] = df["Born"].str.extract(r'(\(\w{3}\))\s*$')
print(df[["name", "NOC"]].head())
```

```python
# add died date column
df["died_date"] = df["Died"].str.extract(date_pattern)
# convert to date time object
df["died_date"] = pd.to_datetime(df["died_date"], errors="coerce")
```

```python
# df.head()
df.info()
```

```python
# get city region and ountry from born column
location_pattern = r'in\s([\w\s-]+),\s([\s\w-]+)\s(\(\w{3}\))'
df[["born_city","born_region", "born_country"]] = df["Born"].str.extract(location_pattern, expand=True)
```
