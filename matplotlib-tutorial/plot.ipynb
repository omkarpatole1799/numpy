```python
from matplotlib import pyplot as plt
import numpy as np
```

```python

x_ages = [22,23,24,25,26]
y_dev_salary = [23000, 28000, 32000, 45000, 40000]

# adding python dev salary by age
y_python_dev = [30000, 32000, 45000, 46000, 50000]

# javascript dev salary by age
y_js_dev = [29000, 31000, 37000, 39000,49287]

plt.title("Salary By Age")

plt.xlabel("Ages")
plt.ylabel("Salary (INR)")
```

```python


# plotting all dev salary
plt.plot(x_ages,y_dev_salary,label='All Dev', color="#969696", linestyle="--", marker='.')

# plotting py dev salary
plt.plot(x_ages, y_python_dev, label="Python Dev", color="#4848ce", marker=".")

# plotting js dev salary
plt.plot(x_ages, y_js_dev, label="JS Dev", color="#C4C730A9", marker=".")
plt.grid()
plt.legend()
plt.show()
# plt.bar(x_ages, y_dev_salary, label="All Dev", color="#969696", alpha=0.5)
```

```python
plt.title("Dev Salary")
# plotting all dev salary (bar chart)
plt.bar(x_ages, y_dev_salary, label="All Dev")

# plotting py dev salary (bar chart)
plt.bar(x_ages, y_python_dev, label="Py Dev")

# plotting js dev salary (bar chart)
plt.bar(x_ages, y_js_dev, label="Js Dev")

plt.grid()
plt.legend()
```

```python
# creating vertical bar charts
print(x_ages)

bar_width = 0.2
x_indexes = np.arange( len(x_ages))

plt.title("Dev Salary")
# plotting all dev salary (bar chart)
plt.bar(x_indexes - bar_width, y_dev_salary, label="All Dev", width=bar_width, color="#00ff00")

# plotting py dev salary (bar chart)
plt.bar(x_indexes, y_python_dev, label="Py Dev", width=bar_width, color="#ff0000")

# plotting js dev salary (bar chart)
plt.bar(x_indexes + bar_width, y_js_dev, label="Js Dev", width=bar_width, color="#0000ff")


plt.xticks(ticks=x_indexes, labels=x_ages)
plt.grid()
plt.legend()
```

```python
import pandas as pd


df_bios = pd.read_csv("../data/bios.csv")
df_bios[["name","born_date", "died_date"]].head()
```

```python
# # df_bios.info()

df_bios["born_date"] = pd.to_datetime(df_bios["born_date"])
df_bios["died_date"] = pd.to_datetime(df_bios["died_date"])


df_bios["age"] = df_bios["died_date"].dt.year - df_bios['born_date'].dt.year 



df_bios_cleaned = df_bios[["name", "height_cm", "weight_kg", "age"]].copy()
```

```python
df_bios_cleaned = df_bios_cleaned[
df_bios_cleaned["age"].notna() & df_bios_cleaned.height_cm.notna() & df_bios_cleaned.weight_kg.notna()]
```

```python
plot_df = df_bios_cleaned.groupby("age")["height_cm"].mean().reset_index()
plot_df_weight = df_bios_cleaned.groupby("age")["weight_kg"].mean().reset_index()
```

```python
print(plot_df.head())
print(plot_df_weight.head())
```

```python
plt.plot(plot_df["age"], plot_df["height_cm"], label="Age VS height (cm)")
# plt.plot(plot_df_weight["age"], plot_df_weight['weight_kg'], label="Age VS Weight (kg)")
plt.xlabel("Age")
plt.ylabel("Height (cm)")
plt.legend()
plt.grid()
```
