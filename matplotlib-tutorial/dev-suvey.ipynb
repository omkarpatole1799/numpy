```python
from matplotlib import pyplot as plt
import numpy as np
import pandas as pd
```

```python
df_data = pd.read_csv('./data/data.csv',)
```

```python
df_data.head()

from collections import Counter
# getting the language counts

ids = df_data['Responder_id']
lang_response = df_data['LanguagesWorkedWith']

lang_counter = Counter()

for resp in lang_response:
  lang_counter.update(resp.split(';'))


y_languages = []
x_popularity = []

for item in lang_counter.most_common(20):
  x_popularity.append(item[1])
  y_languages.append(item[0])

y_languages.reverse()
x_popularity.reverse()

  
```

```python



plt.barh(y_languages, x_popularity, color="#fa02fa", label="Popularity")
plt.xlabel("Popularity")
plt.ylabel("Languages")
plt.title("Most popular programming languages")



plt.legend()

plt.show()
```
